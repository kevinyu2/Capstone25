```python
import pandas as pd
import numpy as np
```

```python
expression_df = pd.read_csv("../data/sct_scale_HT268B1-Th1H3.tsv", sep = '\t', index_col = 0)
expression_raw_df = pd.read_csv("../data/expression_counts_HT268B1-Th1H3.tsv", sep = '\t', index_col = 0)

count_df = pd.read_csv("../data/cnaObj_HT268B1-Th1H3.tsv", sep = '\t', index_col = 0)
```

```python
# This specific one has different names, change if not the case!

count_df.index = count_df.index.str.split('#').str[-1]
expression_df.index = expression_df.index.str.split('Th1H3_').str[-1]
expression_raw_df.index = expression_raw_df.index.str.split('Th1H3_').str[-1]
```

```python
genes_expression = expression_df.columns.to_list()
genes_count = count_df.columns.to_list()

cells_expression = expression_df.index.to_list()
# cells_expression_temp = []
# for cell in cells_expression :
#     expression_df.index = expression_df.index.str.split('_').str[-1]
#     cells_expression_temp.append(cell.split('_')[-1])
# cells_expression = cells_expression_temp

cells_count = count_df.index.to_list()
```

```python
common_genes = list(set(genes_expression) & set(genes_count))
common_cells = list(set(cells_expression) & set(cells_count))
```

```python
print(f"Num cells in expression: {len(cells_expression)}, Num cells in count: {len(cells_count)}")
print(f"Num cells in both: {len(common_cells)}")

print(f"Num genes in expression: {len(genes_expression)}, Num genes in count: {len(genes_count)}")
print(f"Num genes in both: {len(common_genes)}")
```

```python
count_df_final = count_df.loc[common_cells]
expression_df_final = expression_df.loc[common_cells]
expression_raw_df_final = expression_raw_df.loc[common_cells]

```

```python
count_df_final.to_csv("../data/count_matrix_HT268B1-Th1H3.tsv", sep = '\t')
expression_df_final.to_csv("../data/expression_matrix_HT268B1-Th1H3.tsv", sep = '\t')
expression_raw_df_final.to_csv("../data/expression_raw_matrix_HT268B1-Th1H3.tsv", sep = '\t')
```
